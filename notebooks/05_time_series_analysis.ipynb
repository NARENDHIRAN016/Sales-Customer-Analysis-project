# 05 Time Series Analysis

This notebook performs time series analysis on sales data.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/cleaned_sales_data.csv")
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df = df.set_index("Order_Date")

# Monthly Sales Trend
monthly_sales = df.resample("M")["Sales"].sum()
monthly_sales.plot()
plt.title("Monthly Sales Trend")
plt.savefig("visuals/charts/monthly_sales_trend.png")
plt.show()

# Simple Forecast (Moving Average)
monthly_sales.rolling(3).mean().plot(label="3-Month Moving Avg")
plt.legend()
plt.title("Monthly Sales with 3-Month Moving Average")
plt.savefig("visuals/charts/monthly_sales_moving_average.png")
plt.show()
```
